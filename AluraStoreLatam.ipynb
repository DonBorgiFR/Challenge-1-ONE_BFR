### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

###Evaluación de:
· Facturación total de cada tienda

· Categorías más populares

. Promedio de calificación de los clientes

· Productos más y menos vendidos

· Costo promedio del envío

#1. Análisis de facturación



```python
tienda['Tienda'] = 'Tienda 1'
tienda2['Tienda'] = 'Tienda 2'
tienda3['Tienda'] = 'Tienda 3'
tienda4['Tienda'] = 'Tienda 4'
```

Unimos DataFrames en uno solo

```python
df_combinado = pd.concat([tienda, tienda2, tienda3, tienda4], ignore_index=True)
```

Facturación por tienda

```python
print("\nFacturación total por tienda:")

facturacion_por_tienda = df_combinado.groupby('Tienda')['Precio'].sum()
facturacion_formateada = facturacion_por_tienda.round(0).apply(lambda x: f"{int(x):,}".replace(",", "."))

print(facturacion_formateada)
```

# 2. Ventas por categoría

```python
print(df_combinado.columns)
```

```python
print("\nFacturación promedio por venta en cada categoría:")

promedio_por_categoria = df_combinado.groupby('Categoría del Producto')['Precio'].mean()

# Redondea y da formato con separador de miles
promedio_formateado = promedio_por_categoria.round(0).apply(lambda x: f"{int(x):,}".replace(",", "."))

print(promedio_formateado)
```

# 3. Calificación promedio de la tienda


```python
print("\nCalificación promedio por tienda con estrellas:")

calificacion_promedio = df_combinado.groupby('Tienda')['Calificación'].mean().round(2)

for tienda, calificacion in calificacion_promedio.items():
    estrellas = "⭐" * int(round(calificacion))
    print(f"{tienda}: {calificacion} {estrellas}")
```

# 4. Productos más y menos vendidos

```python
print("\nProducto más vendido:")
producto_mas_vendido = df_combinado['Producto'].value_counts().idxmax()
ventas_max = df_combinado['Producto'].value_counts().max()
print(f"{producto_mas_vendido} ({ventas_max} ventas)")

print("\nProducto menos vendido:")
producto_menos_vendido = df_combinado['Producto'].value_counts().idxmin()
ventas_min = df_combinado['Producto'].value_counts().min()
print(f"{producto_menos_vendido} ({ventas_min} venta)")
```

Por tienda

```python
print("\nProducto más vendido por tienda:")

# Agrupamos por 'Tienda' y luego en cada grupo contamos los productos
productos_por_tienda = df_combinado.groupby('Tienda')['Producto'].value_counts()

# Ahora, para cada tienda (nivel 0 del índice), encontramos el índice (Producto) con el máximo conteo.
# El resultado de idxmax() aquí será una Serie con Tienda como índice y una tupla (Tienda, Producto) como valor
producto_mas_vendido_info = productos_por_tienda.groupby(level=0).idxmax()

# También obtenemos la cantidad máxima vendida para cada tienda
cantidad_mas_vendida = productos_por_tienda.groupby(level=0).max()

# Creamos el DataFrame resumen extrayendo el nombre del producto de la tupla
# Usamos .apply(lambda x: x[1]) para tomar el segundo elemento de cada tupla
resumen_mas_vendido_por_tienda = pd.DataFrame({
    'Producto Más Vendido': producto_mas_vendido_info.apply(lambda x: x[1]),
    'Cantidad Vendida': cantidad_mas_vendida
})

print(resumen_mas_vendido_por_tienda)
```

El más vendido por tienda y categoría

```python
print("\nProducto más vendido por tienda y categoría:")

# Agrupar por Tienda y Categoría del Producto, luego contar los productos
productos_por_tienda_categoria = df_combinado.groupby(['Tienda', 'Categoría del Producto'])['Producto'].value_counts()

# El resultado es una Serie con un índice de tres niveles: (Tienda, Categoría del Producto, Producto)
# Para encontrar el producto más vendido dentro de cada combinación de Tienda y Categoría,
# agrupamos por los primeros dos niveles del índice y usamos idxmax()
producto_mas_vendido_tienda_categoria_info = productos_por_tienda_categoria.groupby(level=[0, 1]).idxmax()

# Obtenemos la cantidad vendida de esos productos
cantidad_mas_vendida_tienda_categoria = productos_por_tienda_categoria.groupby(level=[0, 1]).max()

# Ahora creamos un DataFrame resumen. La información del producto y la categoría
# está en el índice de producto_mas_vendido_tienda_categoria_info.
# Necesitamos extraer el nombre del producto (el tercer nivel del MultiIndex resultante de idxmax)
# y mantener la Tienda y Categoría como índice para la claridad.

# product_info.apply(lambda x: x[2]) extrae el nombre del producto de la tupla (Tienda, Categoria, Producto)
resumen_mas_vendido_tienda_categoria = pd.DataFrame({
    'Producto Más Vendido': producto_mas_vendido_tienda_categoria_info.apply(lambda x: x[2]),
    'Cantidad Vendida': cantidad_mas_vendida_tienda_categoria
})

print(resumen_mas_vendido_tienda_categoria)
```

```python
print("\nCosto promedio de envío por tienda:")

# Agrupar por Tienda y calcular el promedio del Costo de envío
costo_envio_promedio_por_tienda = df_combinado.groupby('Tienda')['Costo de envío'].mean()

# Opcional: Redondear para mejor lectura
print(costo_envio_promedio_por_tienda.round(2))
```
